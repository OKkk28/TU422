## Task 3: Password Strength Checker

```python
password = input("please enter your password")
special_characters = "!@#$%&*"
if len(password) > 8 and password.isdigit() == False and any(a in special_characters for a in password) :
    print("valid password")
else:
    print("your password is not strong")
        
```

## Task 5: Temperature Conversion Table

```python
Celsius = 0
Fahrenheit = 0
while Celsius != ("End"):
    Fahrenheit = int(Celsius) * 9/5 + 32
    print("The Fahrenheit Temperature is",Fahrenheit)
    Celsius = input("what is the temperature in Calsius?")
```

## Task 6: Factorial with Input Validation

```python
n = int(input("Enter a number "))
factorial = 1
i = 1

if n < 0:
    print("error, please enter a non-negative number")
else:
    while i <= n:
        factorial *= i
        i += 1
    print(f"{n}! = {factorial}")
```

## Task 7: Prime Number Checker

```python
n = int(input("Enter a number"))
i = 2
is_prime = True

while i < n:
    if n % i ==0:
        is_prime = False
        break
    i += 1
if is_prime:
    print(n,"is prime number")
else:
    print(n,"is not prime number")
```
