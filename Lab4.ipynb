**1. Reads the `grades.txt` file and parses each line into a dictionary.**

```python
students = []
try:
    with open('grades.txt', 'r') as f:
        for line in f:
            name, grade = line.strip().split(",")
        
            student = {
                "name" : name.strip(),
                "grade" : int(grade)
            }
            students.append(student)
    print(students)
except FileNotFoundError:
    print("'grades.text' is not founded")
except ValueError:
    print("value Error")
```

**2.Calculates and prints the average grade.**

```python
total = 0
for student in students:
    total += student["grade"]
average = total/len(students)
print(average)
```

**3. Saves a new file `above_average.csv` containing only the names and grades of students
who scored above average.**

```python
import csv
with open('above_average.csv','w') as file:
    csvwriter = csv.writer(file)
    csvwriter.writerow(["name","grades"])
    for student in students:
        if student["grade"]> average:
            csvwriter.writerow([student["name"],student["grade"]])
```
